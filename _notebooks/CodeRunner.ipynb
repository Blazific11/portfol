## 1. CPT Concept Map

| Concept | Sports-tracker example | Evidence in a CPT program |
| --- | --- | --- |
| Input | Player name and game statistics | `INPUT()` or form fields |
| Output | Player report and average points | `DISPLAY()` or a webpage update |
| List | A collection of points or rebounds | Stores multiple values |
| Procedure | `calculate_average(stats)` | Parameterized abstraction with a return value |
| Selection | Starter and performance decisions | `IF/ELSE` handles different outcomes |
| Iteration | Process every game statistic | A loop avoids duplicated code |
| Boolean logic | Starter status and scoring thresholds | `AND`, `OR`, and `NOT` combine conditions |

College Board pseudocode lists use **1-based indexing**. The Python examples below use Python's normal 0-based indexing, so do not confuse the two when tracing code on the exam.

### List, procedure, selection, and iteration

```text
PROCEDURE getPerformanceLevel(points)
{
  IF (points >= 25)
  {
    RETURN("All-star")
  }
  ELSE
  {
    IF (points >= 15)
    {
      RETURN("Starter")
    }
    ELSE
    {
      RETURN("Developing")
    }
  }
}

PROCEDURE calculateAverage(stats)
{
  total <- 0
  FOR EACH value IN stats
  {
    total <- total + value
  }
  IF (LENGTH(stats) > 0)
  {
    RETURN(total / LENGTH(stats))
  }
  ELSE
  {
    RETURN(0)
  }
}

gamePoints <- [22, 28, 17, 25]
average <- calculateAverage(gamePoints)
level <- getPerformanceLevel(average)
DISPLAY("Average points: " + average)
DISPLAY("Performance level: " + level)
```

Trace the order: the procedures are defined first, then called. The loop visits every item in `gamePoints`, and selection chooses exactly one performance branch.

````python
# Python translation of the complete sports statistics algorithm.

def get_performance_level(points):
    if points >= 25:
        return "All-star"
    if points >= 15:
        return "Starter"
    return "Developing"


def calculate_average(stats):
    if not stats:
        return 0
    total = 0
    for value in stats:
        total += value
    return total / len(stats)


player_name = "Jordan"
game_points = [22, 28, 17, 25]
average = calculate_average(game_points)
performance_level = get_performance_level(average)

print(f"Player: {player_name}")
print(f"Game points: {game_points}")
print(f"Average points: {average:.1f}")
print(f"Performance level: {performance_level}")
````

````python
# List operations and Boolean logic.

def find_player_stat(players, stat_values, target_name):
    for index, player in enumerate(players):
        if player == target_name:
            return stat_values[index]
    return None


players = ["Jordan", "Casey", "Riley", "Morgan"]
points_per_game = [23.0, 18.5, 21.0, 12.5]
search_name = "Riley"
found_average = find_player_stat(players, points_per_game, search_name)

if found_average is not None and found_average >= 20:
    message = f"{search_name} is an elite scorer with {found_average} points per game."
elif found_average is not None:
    message = f"{search_name} is contributing {found_average} points per game."
else:
    message = f"{search_name} was not found."

print(message)
````

## 3. JavaScript Browser Prototype

This prototype maps the same concepts to browser input and output. Save the snippet as an HTML file to run it in a browser.

```html
<label>
  Player name
  <input id="playerName" value="Jordan">
</label>
<label>
  Points by game, separated by commas
  <input id="statInput" value="22, 28, 17, 25">
</label>
<button id="reportButton">Create player report</button>
<pre id="output"></pre>

<script>
function getPerformanceLevel(points) {
  if (points >= 25) return "All-star";
  if (points >= 15) return "Starter";
  return "Developing";
}

function calculateAverage(stats) {
  if (stats.length === 0) return 0;
  let total = 0;
  for (const value of stats) total += value;
  return total / stats.length;
}

document.getElementById("reportButton").addEventListener("click", () => {
  const name = document.getElementById("playerName").value.trim();
  const pointsByGame = document.getElementById("statInput").value
    .split(",")
    .map(value => Number(value.trim()))
    .filter(value => Number.isFinite(value) && value >= 0);
  const average = calculateAverage(pointsByGame);
  document.getElementById("output").textContent =
    `Player: ${name}\nAverage points: ${average.toFixed(1)}\nPerformance: ${getPerformanceLevel(average)}`;
});
</script>
```

**Compare:** JavaScript reads values from DOM elements and writes to the page. Python uses `input()` and `print()`. Both versions use arrays/lists, functions, selection, and iteration.

## 4. Full-Stack Architecture

```mermaid

  A[Frontend: HTML CSS JavaScript] -->|POST /api/stats| B[Backend: Python API]
```

### Responsibilities

| Layer | Responsibilities | CPT concepts |
| --- | --- | --- |
| Frontend | Collect player names, game statistics, and searches; validate fields; display reports | Input, output, selection |
| Backend | Validate requests, calculate averages, search players, enforce roles | Procedures, selection, iteration, Boolean logic |
| Database | Persist teams, players, games, and statistics across sessions | Lists/collections at a larger scale |

Example endpoint:

```python
@app.post("/api/stats")
def add_stat():
    data = request.json
    player_name = data["player"]
    points = data["points"]
    db.execute(
        "INSERT INTO player_stats (player_name, points) VALUES (?, ?)",
        (player_name, points),
    )
    return {"status": "success"}
```

Separating the layers makes the interface easier to change, keeps database access in one trusted location, and allows multiple clients to use the same statistics-processing logic. Authentication belongs in the backend: verify the user before allowing a coach or administrator to edit or delete official statistics.
