## AP CSP CPT concepts in Python and Java

Turn a box score into a player report. This notebook translates the same complete algorithm into Python and Java using input, output, lists, procedures, selection, iteration, and Boolean logic.

**Challenge:** Enter a player name and comma-separated points from several games. The program calculates the average, finds the best game, and assigns a performance tier.

> AP CSP pseudocode lists start at index 1. Python lists and Java collections start at index 0.

````python
# CODE_RUNNER: Run the Python sports statistics report, then edit the player or scores.

def calculate_average(points):
    if len(points) == 0:
        return 0.0

    total = 0
    for score in points:
        total += score
    return total / len(points)


def get_performance_level(average):
    if average >= 25:
        return "All-Star"
    elif average >= 15:
        return "Starter"
    else:
        return "Developing"


player_name = "Jordan"
raw_points = "22, 28, 17, 25"

try:
    game_points = [int(score.strip()) for score in raw_points.split(",")]
except ValueError:
    game_points = []

valid_input = player_name != "" and len(game_points) > 0 and min(game_points) >= 0
if valid_input:
    average = calculate_average(game_points)
    level = get_performance_level(average)
    best_game = max(game_points)
    print(f"{player_name} averages {average:.1f} points")
    print(f"Best game: {best_game} points")
    print(f"Tier: {level}")
else:
    print("Add a player name and at least one valid score.")
````

## Java translation

The Java version uses `ArrayList<Integer>` for the score list and `Scanner` for input. The algorithm is the same: validate, iterate, calculate, classify, and display.

```java
import java.util.ArrayList;
import java.util.Scanner;

public class SportsScoutboard {
    public static double calculateAverage(ArrayList<Integer> points) {
        if (points.isEmpty()) {
            return 0.0;
        }

        int total = 0;
        for (int score : points) {
            total += score;
        }
        return (double) total / points.size();
    }

    public static String getPerformanceLevel(double average) {
        if (average >= 25) {
            return "All-Star";
        } else if (average >= 15) {
            return "Starter";
        } else {
            return "Developing";
        }
    }

    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        System.out.print("Player name: ");
        String playerName = scanner.nextLine().trim();
        System.out.print("Points by game, separated by commas: ");
        String[] rawPoints = scanner.nextLine().split(",");
        ArrayList<Integer> gamePoints = new ArrayList<>();

        try {
            for (String rawPoint : rawPoints) {
                gamePoints.add(Integer.parseInt(rawPoint.trim()));
            }
        } catch (NumberFormatException error) {
            gamePoints.clear();
        }

        boolean validInput = !playerName.isEmpty() && !gamePoints.isEmpty();
        for (int score : gamePoints) {
            if (score < 0) {
                validInput = false;
            }
        }

        if (validInput) {
            double average = calculateAverage(gamePoints);
            String level = getPerformanceLevel(average);
            int bestGame = gamePoints.get(0);
            for (int score : gamePoints) {
                if (score > bestGame) {
                    bestGame = score;
                }
            }
            System.out.printf("%s averages %.1f points%n", playerName, average);
            System.out.println("Best game: " + bestGame + " points");
            System.out.println("Tier: " + level);
        } else {
            System.out.println("Add a player name and at least one valid score.");
        }
        scanner.close();
    }
}
```

## CPT checklist

- Input changes the report.
- Output communicates a meaningful sports result.
- A list stores multiple game values.
- Parameterized procedures are called and their return values are used.
- Iteration and selection work together in one algorithm.
- Empty lists and invalid scores are handled safely.

````python
// CODE_RUNNER: Run the Java sports statistics report, then edit the player or scores.

import java.util.ArrayList;
import java.util.Arrays;

public class Main {
    public static double calculateAverage(ArrayList<Integer> points) {
        if (points.isEmpty()) {
            return 0.0;
        }

        int total = 0;
        for (int score : points) {
            total += score;
        }
        return (double) total / points.size();
    }

    public static String getPerformanceLevel(double average) {
        if (average >= 25) {
            return "All-Star";
        } else if (average >= 15) {
            return "Starter";
        } else {
            return "Developing";
        }
    }

    public static void main(String[] args) {
        String playerName = "Jordan";
        ArrayList<Integer> gamePoints = new ArrayList<>(Arrays.asList(22, 28, 17, 25));
        boolean validInput = !playerName.isEmpty() && !gamePoints.isEmpty();

        for (int score : gamePoints) {
            if (score < 0) {
                validInput = false;
            }
        }

        if (validInput) {
            double average = calculateAverage(gamePoints);
            String level = getPerformanceLevel(average);
            int bestGame = gamePoints.get(0);
            for (int score : gamePoints) {
                if (score > bestGame) {
                    bestGame = score;
                }
            }
            System.out.printf("%s averages %.1f points%n", playerName, average);
            System.out.println("Best game: " + bestGame + " points");
            System.out.println("Tier: " + level);
        } else {
            System.out.println("Add a player name and at least one valid score.");
        }
    }
}

Main.main(null);
````
