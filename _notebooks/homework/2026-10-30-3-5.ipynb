### Search filter

Prediction: 1.2 matches one record, flywheel matches 1.1 and 2.1, absent matches none. Invalid categories are excluded even when the name matches.

````python
# CODE_RUNNER: Search filter
records = [
    {"product_name":"Replacement Flywheels and Clutch Assemblies","category":"Auto Racing","spec_number":"1.1"},
    {"product_name":"Multiple Disc Clutch Assemblies","category":"Drag Racing","spec_number":"1.2"},
    {"product_name":"Racing Flywheel Record","category":"Auto Racing","spec_number":"2.1"},
    {"product_name":"Flywheel","category":"Street Car","spec_number":"3.1"}]
for query in ["1.2", "flywheel", "absent"]:
    matches = []
    for record in records:
        name_match = query.lower() in record["product_name"].lower()
        spec_match = query == record["spec_number"]
        category_ok = record["category"] in ["Auto Racing", "Drag Racing"]
        if (name_match or spec_match) and category_ok:
            matches.append(record["spec_number"])
    print(query, matches)
````

### Reusable SFI validator

Prediction: Accept 1.2; reject Street Car and duplicate 1.1. Empty required fields also fail.

````python
# CODE_RUNNER: Reusable SFI validator
def valid_record(part, existing):
    has_name = isinstance(part.get("product_name"), str) and bool(part["product_name"].strip())
    has_spec = isinstance(part.get("spec_number"), str) and bool(part["spec_number"].strip())
    has_date = isinstance(part.get("effective_date"), str) and bool(part["effective_date"].strip())
    valid_category = part.get("category") in ["Auto Racing", "Drag Racing"]
    duplicate_spec = part.get("spec_number") in existing
    return has_name and has_spec and has_date and valid_category and not duplicate_spec

existing_spec_numbers = ["1.1", "2.1"]
test_records = [
 {"product_name":"Multiple Disc Clutch Assemblies","category":"Auto Racing","spec_number":"1.2","effective_date":"Feb. 9, 2006"},
 {"product_name":"Replacement Flywheels","category":"Street Car","spec_number":"3.1","effective_date":"Jan. 1, 2026"},
 {"product_name":"Existing Flywheel Record","category":"Drag Racing","spec_number":"1.1","effective_date":"Jan. 1, 2026"}]
for record, expected in zip(test_records, [True, False, False]):
    accepted = valid_record(record, existing_spec_numbers)
    print(record["spec_number"], "ACCEPT RECORD" if accepted else "REJECT RECORD")
    assert accepted == expected
for field in ["product_name", "spec_number", "effective_date"]:
    bad = dict(test_records[0])
    bad[field] = " "
    assert not valid_record(bad, existing_spec_numbers)
    print("Whitespace-only", field, "rejected")
````

```text
Whitespace-only product_name rejected
Whitespace-only spec_number rejected

```
