## Popcorn

Python index 1 and AP index 2 select the second level. Replacement preserves length; append adds one element.

### Extend the level sequence

Prediction: [6, 12, 12, 5] and 4; changed append gives [6, 12, 12, 8] and 4.

````python
# CODE_RUNNER: Extend the level sequence
for added in [5, 8]:
    level_stars = [6, 9, 12]
    level_stars[1] = 12
    level_stars.append(added)
    print(level_stars)
    print(len(level_stars))
````

## Homework

### Grow the game and test an empty project

Prediction: [12, 23, 10, 7], 4 levels, 52 stars; empty project has 0 levels and 0 stars.

````python
# CODE_RUNNER: Grow the game and test an empty project
def report(level_stars):
    total_stars = 0
    for stars in level_stars:
        total_stars = total_stars + stars
    print(level_stars)
    print("Level count:", len(level_stars))
    print("Total stars:", total_stars)
    return total_stars
level_stars = [12, 18, 10]
level_stars[1] = level_stars[1] + 5
level_stars.append(7)
assert report(level_stars) == 52
assert report([]) == 0
````

**Actual output**

```text
[12, 23, 10, 7]
Level count: 4
Total stars: 52
[]
Level count: 0
Total stars: 0
```
