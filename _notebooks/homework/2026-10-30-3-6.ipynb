### Score and advise

Prediction: The inputs reach all four branches at scores 9, 6, 3, 0. Exactly 100 does not trigger the >100 signal.

````python
# CODE_RUNNER: Score and advise
def advise(risk_score):
    # 3 thresholds -> 4 outcomes -> 4 answers
    if risk_score >= 9:
        print("High risk: do not reply; verify through a trusted channel.")
    elif risk_score >= 6:
        print("Risky: check with someone you trust before replying.")
    elif risk_score >= 3:
        print("Caution: verify the request before acting.")
    else:
        print("No listed signals detected; this does not prove the message is safe.")

def score_message(message, amount_requested, known_contact):
    risk_score = 0
    # Each independent if adds an optional contribution; a false signal adds zero.
    # No else is needed here because the final advice always prints an answer.
    if "gift card" in message.lower():
        risk_score += 4
    if amount_requested > 100:
        risk_score += 3
    if not known_contact:
        risk_score += 2
    print("Score:", risk_score)
    advise(risk_score)
    return risk_score
tests = [
 ("URGENT: send gift cards now", 500, False, 9),
 ("Please send gift cards", 0, False, 6),
 ("Please send the payment", 101, True, 3),
 ("See you Sunday", 0, True, 0),
 ("Payment request", 100, True, 0)]
for message, amount, known, expected in tests:
    assert score_message(message, amount, known) == expected
````
