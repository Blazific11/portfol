## Homework

`game_title` is text, `reduced_motion` is Boolean, and level counts are integers. Python `=` assigns; `==` compares.

### Star counter and both fresh tests

Prediction: Bonus 5 produces 40, 40, 45; bonus 0 produces 40, 40, 40.

````python
# CODE_RUNNER: Star counter and both fresh tests
def star_counter(bonus):
    game_title = "Orbit Star Trail"
    reduced_motion = True
    level_one_stars, level_two_stars, level_three_stars = 12, 18, 10
    total_stars = level_one_stars + level_two_stars + level_three_stars
    print(game_title)
    print("Reduced motion:", reduced_motion)
    print("Original total:", total_stars)
    level_two_stars = level_two_stars + bonus
    print("Stored total:", total_stars)
    total_stars = level_one_stars + level_two_stars + level_three_stars
    print("Recalculated total:", total_stars)
    return total_stars
for bonus, expected in [(5, 45), (0, 40)]:
    print("Bonus:", bonus)
    assert star_counter(bonus) == expected
````

**Actual output**

```text
Recalculated total: 45
Bonus: 0
Orbit Star Trail
Reduced motion: True
Original total: 40
Stored total: 40
Recalculated total: 40
```

## Tests

Both cases initialize fresh state. The assertions passed. A total stores an evaluated result, not a formula linked to its inputs.

## Design Thinking

A player may prefer less animation. The goal is understandable progress with a player-controlled reduced-motion setting. Separate numeric counts make reassignment easy to trace; the prototype preserves a checkpoint and recomputes the total. Testing bonus zero checks that recalculation does not invent stars. Reduced motion should be a choice because players have different comfort and accessibility needs. Peer playtest feedback has not been supplied.
